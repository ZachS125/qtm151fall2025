```python

import pandas as pd
import numpy as np
pd.set_option("display.precision", 2)
```

## 📦 Datasets (created for you)

```python

# --- Dataset 1: Movies ---
titles = [
    "Past Lives", "Citizen Kane", "Revenge", "Okja", "Shutter",
    "A Tale of Two Sisters", "La Chinoise", "The Umbrellas of Cherbourg",
    "House", "Blue Velvet"
]

# Note the countries include values we may want to replace
countries = [
    "American", "American", "France", "Korean", "Thailand",
    "Korean", "France", "France", "Japan", "British"
]

runtime = [106, 119, 108, 120, 97, 115, 96, 91, 88, 120]
year =     [2023, 1941, 2017, 2017, 2004, 2003, 1967, 1964, 1977, 1986]

df_movies = pd.DataFrame({
    "title": titles,
    "country": countries,
    "runtime": runtime,
    "year": year
})

# --- Dataset 2: Scores (to be merged later) ---
scores = [7.8, 8.2, 6.4, 7.3, 7.0, 7.1, 6.9, 7.8, 7.2, 7.7]
directors = [
    "Celine Song", "Orson Welles", "Coralie Fargeat", "Bong Joon Ho",
    "Banjong Pisanthanakun", "Kim Jee-woon", "Jean-Luc Godard",
    "Jacques Demy", "Nobuhiko Obayashi", "David Lynch"
]

df_scores = pd.DataFrame({
    "title": titles,
    "imdb_score": scores,
    "director": directors
})

# --- Dataset 3: Sales ---
# Synthetic sales dataset for grouping/aggregation with a filter step
rng = np.random.default_rng(42)
n = 120
regions = rng.choice(["North", "South", "East", "West"], size=n)
products = rng.choice(["Widget", "Gadget", "Doohickey", "Thingamajig"], size=n)
units = rng.integers(1, 50, size=n)
unit_price = rng.choice([9.99, 14.99, 19.99, 29.99], size=n)
dates = pd.date_range("2021-01-01", periods=n, freq="D")
df_sales = pd.DataFrame({
    "order_id": np.arange(1, n+1),
    "date": dates,
    "region": regions,
    "product": products,
    "units": units,
    "unit_price": unit_price
})
df_sales["revenue"] = (df_sales["units"] * df_sales["unit_price"]).round(2)

print("df_movies (head):")
display(df_movies.head())
print("\ndf_scores (head):")
display(df_scores.head())
print("\ndf_sales (head):")
display(df_sales.head())
```

## 📝 Questions

### **Q1. Replace values with `.replace` (use `df_movies`)**
In the `country` column, replace:
- `"American"` → `"USA"`  
- `"Korean"` → `"ROK"`  
- `"British"` → `"UK"`

Do this with a **single call** to `.replace`. Show the updated `df_movies["country"]`.


```python

# Q1 — YOUR CODE HERE
# Replace 3 values in df_movies['country'] using .replace(...)
# Show df_movies['country'] at the end.
df_movies["country_changed"] = df_movies["country"].replace(["American", "Korean", "British"], ["USA", "ROK", "UK"])
print(df_movies["country_changed"])
print(df_movies["country"])
```


### **Q2. Filter → Group → Aggregate → Filter (use `df_sales`)**
- First, subset to rows where `region == "West"`
- Then group by `product`
- Aggregate:
  - mean revenue (`mean_revenue`) from `"revenue"`
  - total units (`total_units`) from `"units"`
- Finally, **keep only** the groups where `mean_revenue >= 400`

✅ **Full credit** if you do this in **one line with chaining**. Print the result.


```python

# Q2 — YOUR CODE HERE

df_sales_agg = (df_sales.query("region=='West'")
                .groupby("product")
                .agg(
                    mean_revenue = ('revenue', 'mean'), 
                     total_units = ('units', 'sum')
                     )
                     .query("mean_revenue>=400"))
print(df_sales_agg)
```


### **Q3. Recode a numeric variable into categories (use `df_movies`)**
Create a new column `year_brackets` that bins `year` into:
- `"1941–1967"` for years 1941–1967
- `"1968–1999"` for years 1968–1999
- `"2000–onwards"` for years 2000 and above

Use `pd.cut` with `bins=[1940, 1967, 1999, np.inf]` and the labels above.


```python

# Q3 — YOUR CODE HERE
labels = ["1941-1967", "1968-1999", "2000-onwards"]
bins = [1939, 1967, 1999, np.inf]
year_brackets = pd.cut(df_movies["year"], 
                       bins = bins, 
                       right = True, 
                       labels = labels)
print(year_brackets)
```


### **Q4. Group, aggregate, and sort (use `df_sales`)**
Group `df_sales` by `region`, compute for each region:
- `total_revenue` = sum of `revenue`
- `avg_units` = mean of `units`

Then **sort by `total_revenue` descending**.  
✅ Full credit for a **single chained line**. Print the result.


```python

# Q4 — YOUR CODE HERE
region_agg_sorted = (
    df_sales
    .groupby("region")
    .agg(total_revenue=("revenue", "sum"), 
         avg_units=("units", "mean"))
    .sort_values(by="total_revenue", ascending=False)
    .round(2)
)
region_agg_sorted
```


### **Q5. Rename one column (use `df_movies`)**
Rename the `runtime` column to `runtime(minutes)` **without changing other columns**.
Show the new columns list to confirm.


```python

# Q5 — YOUR CODE HERE
df_movies.rename(columns={"runtime": "runtime(minutes)"}, inplace=True)
print(df_movies.columns.values)
```


### **Q6. Merge two DataFrames (`df_movies` + `df_scores`)**
Use `pd.merge` to join `df_movies` (left/primary) with **only** the `title` and `imdb_score`
columns from `df_scores` (right/secondary) on `title`, using a **left join**.

Print the merged DataFrame.


```python

# Q6 — YOUR CODE HERE
merged = pd.merge(
    df_movies,
    df_scores[["title", "imdb_score"]],
    on="title",
    how="left"
)
print(merged)

```
